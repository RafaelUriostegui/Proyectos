###Librerías necesarias

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline
from scipy.interpolate import make_interp_spline
```

### Interpolación spline cúbica directa

```python
# Datos conocidos (pocos puntos)
x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 1, 0, -1, 0])

# 1. Creamos el objeto spline cúbico
spline = CubicSpline(x, y)

# 2. Evaluamos en nuevos puntos más densos para graficar una curva suave
x_nuevo = np.linspace(0, 4, 100)
y_interpolado = spline(x_nuevo)
```

```python
plt.plot(x_nuevo,y_interpolado)
plt.plot(x,y,'o')
plt.title('Spline')
plt.show()
```

### Interpolación spline de orden variable

```python
# 1. Definimos los puntos originales (pocos datos)
x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 0, -2, 0])

# 2. Generamos puntos más densos para evaluar y graficar las curvas continuas
x_nuevo = np.linspace(0, 4, 200)

# 3. Creamos un spline CUADRÁTICO (Grado 2)
spline_2 = make_interp_spline(x, y, k=2)
y_cuadratico = spline_2(x_nuevo)

# 4. Creamos un spline LINEAL (Grado 1) para notar la diferencia
spline_1 = make_interp_spline(x, y, k=1)
y_lineal = spline_1(x_nuevo)

# --- Código opcional para graficar y ver el resultado ---
plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='red', zorder=5, label='Puntos originales')
plt.plot(x_nuevo, y_cuadratico, label='Spline Cuadrático (k=2)', color='blue', linewidth=2)
plt.plot(x_nuevo, y_lineal, label='Spline Lineal (k=1)', color='orange', linestyle='--')

plt.title('Comparación de Órdenes de Interpolación Spline')
plt.legend()
plt.grid('on')
plt.show()
```
